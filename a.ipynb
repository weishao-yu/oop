只要看到等號左邊有加上括號 []（像是 M[0][0] = ... 或 M[0][:] = ...），這通常代表著**「深入修改（In-place modification）」。
如果等號左邊是一個乾淨的變數名稱或第一層的索引（像是 M = ... 或 M[0] = ...），這代表著「重新綁定位址（Rebinding/Reassignment）」**！

```python
M = [[1]*3]*2
print(M)
M[0][0], M[0][1], M[0][2] = 2, 2, 2
print(M)
```

```python
M = [[1]*3]*2
print(M)
M[0][:] = [2]*3
print(M)
```

```python
M = [[1]*3]*2
print(M)
M[0] = [2]*3
print(M)
```

```python
def prime_check(n):
    if n < 2:
        return False
    elif n == 2:
        return True
    else:
        for factor in range(2, int(n**0.5) + 1):
            if n % factor == 0:
                return False
        return True
    
print(prime_check(0))
print(prime_check(1))
print(prime_check(2))
print(prime_check(3))
print(prime_check(89))
print(prime_check(91))
print(prime_check(97))
print(prime_check(2047))
```

```python
def prime_count(N):
    cnt = 0
    for num in range(2,N+1):
        if prime_check(num):
            cnt += 1
    return cnt

print(prime_count(100))
print(prime_count(1000))
print(prime_count(10000))
```

```python
gender_set = list(set(gender_list))
print(gender_set)
```

```python
gender_cnt = { key:0 for key in gender_set }
print(gender_cnt)
for gender in gender_list:
    gender_cnt[gender] += 1
print(gender_cnt)
```

```python
gender_cnt_ = { key:gender_list.count(key) for key in gender_set}
print(gender_cnt_)
```
